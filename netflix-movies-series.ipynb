IQR(Inter Quartilerange) method is used to detect and handle outliers

```python
Q1 = df["duration_number"].quantile(0.25)
Q3 = df["duration_number"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

less than lowerbond
greater than upperbond is called as outliers

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

Feature scaling is applied to the numerical columns `release_year` and `duration_number`.
**StandardScaler:** Standardizes the features so that they have a mean close to 0 and a standard deviation close to 1.
- **MinMaxScaler:** Transforms the feature values to a range between 0 and 1.


```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler

scale_cols = ['release_year', 'duration_number']

standard_scaled = StandardScaler().fit_transform(df[scale_cols])

normalized_scaled = MinMaxScaler().fit_transform(df[scale_cols])

comparison = pd.DataFrame({
    'release_year_raw': df['release_year'].values[:5],
    'release_year_standardized': standard_scaled[:5, 0],
    'release_year_normalized': normalized_scaled[:5, 0],
    
    'duration_raw': df['duration_number'].values[:5],
    'duration_standardized': standard_scaled[:5, 1],
    'duration_normalized': normalized_scaled[:5, 1]
})

comparison
```

The original numerical values are compared with their standardized and normalized versions.

This comparison demonstrates how StandardScaler and MinMaxScaler transform the same numerical features using different approaches.
